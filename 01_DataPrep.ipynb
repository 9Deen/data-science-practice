```python
reviews.drop(['designation','region_1','region_2','taster_name','taster_twitter_handle'],
              axis=1,
              inplace=True
            )
reviews.head()
```

```python
reviews.price.fillna(reviews.price.mean(),inplace=True)
reviews.info()
```

```python
reviews.country.fillna(reviews.country.mode()[0],inplace=True)
reviews.info()
```

```python
reviews.province.fillna(reviews.province.mode()[0],inplace=True)
reviews.info()
```
